```python
import matplotlib.colors as colors
import matplotlib.cm as cm

# ALGOS = ["DE", "DiagonalCMA", "modcma", "modde", "NaiveIsoEMNA", "NGOpt14", "NGOpt38", "OnePlusOne", "PSO", "RandomSearch", "RCobyla"]
# DIMS =  [5, 20]
# BUDGETS = [100, 250, 500, 1000, 2500, 5000, 10000]

ALGOS = ["RCobyla"]
DIMS =  [5, 20]
BUDGETS = [100, 500, 1000, 5000, 10000]

os.makedirs("plots/Images_progress", exist_ok=True)

for SC in ALGOS:
    for DIM in DIMS:
        for BUDGET in BUDGETS:
            print(SC, DIM, BUDGET)
            data, _, sat_ls = load_data_continuous.get_data(
                algo=SC,
                dim=DIM,
                budget=BUDGET,
                runs=50,
                agregate_runs=False,
                log_target=False
            )
            print(data.mean())
            dist_name = "lognorm_floc"
            ann_name = "LOG"
            
            fig, ax = plt.subplots(figsize=(3,3))
            for idx, i in enumerate(range(data.shape[0])): #data.shape[0]):
                sorted = np.sort(data[i, :])
                
                yvals = np.arange(1,len(sorted)+1)/float(len(sorted))
                cval = next(blue_colors)
                ax.step(sorted, yvals, c=cval, linewidth=2)

            ax.tick_params(
            axis='both',       # changes apply to the x-axis
            which='both',      # both major and minor ticks are affected
            top='off',         # ticks along the top edge are off
            labelleft='off',
            labelsize=10)
                
            ax.set_xlabel("regret", fontsize=10)
            ax.set_ylabel(f"CDF (Budget {BUDGET})", fontsize=10)
            # ax.set_xlim(limit_dict[SC])
            ax.set_xscale('log')
            
            plt.tight_layout()
            plt.savefig(f"plots/Images_progress/CDFs_{SC}_{DIM}_{BUDGET}.png")
            plt.show()
```
