### Plot different distributions

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, lognorm, expon, invgauss, weibull_min

# Runtime values
x = np.linspace(0.001, 3, 1000)

# Distribution parameters
mu = 1 # Mean for normal and lognormal
sigma = 0.5 # Standard deviation for normal and lognormal
lambda_ = 1 # Scale parameter for exponential and inverse Gaussian
k = 0.5  # Shape parameter for Weibull distribution

# Distributions
normal_pdf = norm.pdf(x, loc=mu, scale=sigma)
lognorm_pdf = lognorm.pdf(x, s=mu, scale=sigma)
exp_pdf = expon.pdf(x, scale=1/lambda_)
invgauss_pdf = invgauss.pdf(x, mu=mu, scale=1/lambda_)
weibull_min_pdf = weibull_min.pdf(x, c=1/k, scale=1/lambda_) 
plt.rcParams.update({
    "font.size": 16,      # default text size
    "axes.titlesize": 25, # subplot titles
    "axes.labelsize": 22, # x/y labels
    "xtick.labelsize": 20,
    "ytick.labelsize": 20,
    "legend.fontsize": 18
})
# Plotting
plt.figure(figsize=(10, 7))
plt.plot(x, normal_pdf, label='normal (μ=1, σ=0.5)')
plt.plot(x, lognorm_pdf, label='lognormal (μ=1, σ=0.5)')
plt.plot(x, exp_pdf, label='exponential (λ=1)')
plt.plot(x, invgauss_pdf, label='inverse Gaussian (μ=1, σ=1)')
plt.plot(x, weibull_min_pdf, label='Weibull (λ=1, k=0.5)', color='black')

# plt.title("Five Considered Distribution Families")
plt.xlabel('Algorithm Performance')
plt.ylabel('Probability Density')
plt.legend()
plt.tight_layout()

# Optional: Save figure
plt.savefig("dist_families_plot.pdf")
plt.savefig("dist_families_plot.png")

plt.show()
```
